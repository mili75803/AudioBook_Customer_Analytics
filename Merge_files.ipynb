```python
import pandas as pd

lr = pd.read_csv("logistic_model_metrics.csv")
nn = pd.read_csv("NN_model_metrics.csv")

# metrics = pd.concat([lr, nn], ignore_index=True)

# metrics.to_csv("model_metrics.csv", index=False)

metrics = lr.merge(nn, on="metric")

metrics.to_csv("model_metrics.csv", index=False)



lr = pd.read_csv("logistic_user_predictions.csv")
nn = pd.read_csv("NN_user_predictions.csv")

# predictions = pd.concat([lr, nn], ignore_index=True)

# predictions.to_csv("model_predictions.csv", index=False)


predictions = pd.DataFrame({
    "y_true": lr["y_true"],
    "LR_score": lr["y_score"],
    "LR_pred": lr["y_pred"],
    "NN_score": nn["y_score"],
    "NN_pred": nn["y_pred"]
})

predictions.to_csv("model_predictions.csv", index=False)
```
